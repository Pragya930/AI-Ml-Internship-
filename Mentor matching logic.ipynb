```python
import pandas as pd

data = {
    "Student_ID": ["S1", "S2", "S3", "S4"],
    "Academic_Score": [35, 75, 55, 60],
    "Wellness_Score": [30, 80, 50, 70],
    "Engagement_Score": [50, 90, 60, 40],
    "Cluster": [3, 0, 1, 2]
}

df = pd.DataFrame(data)
```

```python
def assign_risk(row):
    if row['Academic_Score'] < 40 or row['Wellness_Score'] < 40:
        return "High Risk"
    elif row['Academic_Score'] < 60:
        return "Medium Risk"
    else:
        return "Low Risk"

df['Risk_Level'] = df.apply(assign_risk, axis=1)
```

```python
def recommend_support(cluster):
    if cluster == 0:
        return "Career", "Career Planning & Internships"
    elif cluster == 1:
        return "Academic", "Remedial Classes"
    elif cluster == 2:
        return "Wellness", "Motivation & Counseling"
    elif cluster == 3:
        return "Academic", "Intensive Academic + Wellness Support"

df[['Recommended_Type','Intervention']] = df['Cluster'].apply(
    lambda x: pd.Series(recommend_support(x))
)
```

```python
import pandas as pd


students_data = {
    "Student_ID": ["S1", "S2", "S3", "S4"],
    "Academic_Score": [35, 75, 55, 60],
    "Wellness_Score": [30, 80, 50, 70],
    "Engagement_Score": [50, 90, 60, 40],
    "Cluster": [3, 0, 1, 2],
    "Alert_Status": [2,4,7,2]  
}

df_students = pd.DataFrame(students_data)


mentors_data = {
    "Mentor_ID": ["M1", "M2", "M3", "M4", "M5"],
    "Name": ["Dr. Sharma", "Ms. Kapoor", "Mr. Singh", "Dr. Iyer", "Ms. Rao"],
    "Expertise": ["Academic", "Wellness", "Career", "Academic", "Wellness"],
    "Available_Hours": [10, 8, 6, 12, 10],
    "Max_Students": [5, 4, 6, 8, 5],
    "Mentor_Type": ["Academic Advisor", "Counselor", "Career Coach", "Subject Expert", "Mental Health Mentor"]
}

df_mentors = pd.DataFrame(mentors_data)
```

```python
def assign_risk(row):
    if row['Academic_Score'] < 40 or row['Wellness_Score'] < 40:
        return "High Risk"
    elif row['Academic_Score'] < 60:
        return "Medium Risk"
    else:
        return "Low Risk"

df_students['Risk_Level'] = df_students.apply(assign_risk, axis=1)
```

```python
def recommend_support(cluster):
    if cluster == 0:
        return "Career", "Career Planning & Internships"
    elif cluster == 1:
        return "Academic", "Remedial Classes"
    elif cluster == 2:
        return "Wellness", "Motivation & Counseling"
    elif cluster == 3:
        return "Academic", "Intensive Academic + Wellness Support"

df_students[['Recommended_Type','Intervention']] = df_students['Cluster'].apply(
    lambda x: pd.Series(recommend_support(x))
)
```

```python
def match_mentor(student_type):
    available = df_mentors[df_mentors['Expertise'] == student_type]
    if not available.empty:
        return available.iloc[0]['Name']
    else:
        return "No Mentor Available"

df_students['Assigned_Mentor'] = df_students['Recommended_Type'].apply(match_mentor)
```

```python
def generate_alert(row):
    if row['Risk_Level'] == "High Risk":
        return "🚨 Immediate Intervention Required"
    else:
        return "Normal Monitoring"

df['Alert_Status'] = df.apply(generate_alert, axis=1)
```

```python
final_table = df_students[['Student_ID', 'Cluster', 'Risk_Level',
                           'Recommended_Type', 'Intervention',
                           'Assigned_Mentor', 'Alert_Status' ]]

print(final_table)
```
